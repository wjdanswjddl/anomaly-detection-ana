```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
```

```python
basedir = "/exp/sbnd/app/users/munjung/anomaly-detection/validation/"

with open(basedir + "save_data_img29.pkl", "rb") as f:
# with open(basedir + "save_data_img24.pkl", "rb") as f:
    save_data_wclass = pickle.load(f)

t_key = 2

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

img_orig = save_data_wclass["original"]
img_diffused = save_data_wclass[t_key]["diffused"][0]
img_reconstructed = save_data_wclass[t_key]["reconstructed"][0][0]
# img_reconstructed[img_reconstructed < 0.01] = img_orig[img_reconstructed < 0.01] * 1.5
# img_reconstructed *= 1.4
# img_reconstructed[:, 270:290] *= 1.3
# img_reconstructed[0:100, 0:80] += img_orig[0:100, 0:80]
# img_reconstructed[350:420, 490:500] += img_orig[350:420, 490:500]

ax[0].imshow(img_orig, vmin=-1, vmax=1, cmap="bwr")
ax[0].title.set_text("Original")

ax[1].imshow(img_diffused, vmin=-1, vmax=1, cmap="bwr")
ax[1].title.set_text("Diffused")

ax[2].imshow(img_reconstructed, vmin=-1, vmax=1, cmap="bwr")
ax[2].title.set_text("Reconstructed")

diff = np.abs(img_reconstructed - img_orig)
ax[3].imshow(diff)
ax[3].title.set_text("|Reconstructed - Original|")

# fig.colorbar(ax[2].images[0], ax=ax, orientation='vertical', fraction=0.04)

plt.suptitle("t=200")
plt.show()
```
